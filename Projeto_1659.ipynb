```python
# Os valores estão em Rupia indiana (INR), vamos converter para Dólar americano (USD)
# 1 INR = 0.012 USD (taxa de câmbio aproximada)
cars["selling_price_usd"] = cars["selling_price"] * 0.012
```

```python
cars.drop(columns=["selling_price"], inplace=True)
```

```python
cars.head()
```

```python
# Agora vamos organizar as colunas para ficar mais fácil de trabalhar
cars = cars[["name", "year","selling_price_usd", "km_driven", "fuel", "seller_type", "transmission", "owner"]]
```

```python
cars.selling_price_usd.describe()
```

# 1.Qual a marca de Carros mais comercializada?

```python
cars['name'].nunique()
```

```python
cars['name'].value_counts()
```

```python
cars[cars['name'].str.contains("Maruti")]
```

```python
# Então a Marca mais vendida é a Maruti Suzuki com 1280 carros vendidos.
quantidade = (cars['name'].str.contains('Maruti')).sum()
print(quantidade)
```

# 2.Qual o Modelo, Ano e Valor do carro mais caro vendido?

```python
cars.loc[cars.groupby('year')['selling_price_usd'].idxmax()][['name', 'year', 'selling_price_usd']].sort_values('selling_price_usd', ascending=False).head(1)
```

# 3. Qual é o modelo de carro mais caro vendido em cada ano?

```python
cars.loc[cars.groupby('year')['selling_price_usd'].idxmax()][['name', 'year', 'selling_price_usd']]
```

# 4. Qual modelo de carro teve a maior media de quilometragem (km_driven) entre os carros vendidos?


```python
cars.groupby('name')['km_driven'].mean().sort_values(ascending=False).head(1)
```

# 5. Qual o preço médio do modelo de carro mais comercializado?

```python
cars['name'].value_counts()
```

```python
#Qual o preço médio do modelo de carro mais comercializado?
modelo_mais_comercializado = cars['name'].value_counts().idxmax()
preco_medio = cars[cars['name'] == modelo_mais_comercializado]['selling_price_usd'].mean()
print(f"O modelo mais comercializado é {modelo_mais_comercializado} com um preço médio de {preco_medio:.2f} USD.")
```
